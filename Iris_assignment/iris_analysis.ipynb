# 3章最終課題
# KMeans, Dendrogram, DBSCANを比較して、使い方を考察してください。

## KMeans

KMeans は、あらかじめ決めたクラスタ数にデータを分ける手法であり、各クラスタの重心に近い点どうしを同じ群としてまとめる。計算が比較的単純で結果も分かりやすく、最終的に何群に分けたいかが決まっている場合に使いやすい。一方で、クラスタ数を事前に指定する必要があり、外れ値に弱く、球状でない複雑なクラスタ構造にはあまり向いていない。

## Dendrogram

Dendrogram は階層クラスタリングの結果を木構造として表したものであり、データどうしがどの順番で結合されていくかを可視化できる。クラスタ数を最初に決める必要がなく、後から切る高さを変えて複数の見方ができる点が大きな利点である。その一方で、どこで切るかの判断に主観が入りやすく、サンプル数が多い場合には図が複雑になって読み取りにくくなる。

## DBSCAN

DBSCAN は、点の密度に基づいてクラスタを見つける手法であり、密集している領域をクラスタとし、孤立した点をノイズとして扱う。クラスタ数を事前に決める必要がなく、外れ値を自然に除外できる点が強みである。また、球状でないクラスタにも対応しやすい。しかし、`eps` と `min_samples` の設定に結果が大きく左右され、密度の異なるクラスタが混在する場合にはうまく分けにくい。

## 今回の可視化結果から読み取れること

今回の Iris データでは、KMeans による可視化から、3 クラスタに分けた場合でも 1 群は明確に独立し、残り 2 群はやや近い位置にあることが分かった。Dendrogram でも同様に、1 つの群が他から大きく離れており、残る 2 群は比較的近い距離で結合されることが確認できた。さらに DBSCAN では、密度の観点から見ると独立した 1 群と、それ以外の大きな 1 群、そして少数のノイズ点として表れた。これらを総合すると、Iris データには明確に分かれやすい群が 1 つあり、残る 2 群はある程度似ていて境界があいまいであることが分かる。

## まとめ

以上より、3 手法は同じクラスタリングでも、見ている基準が異なることが分かった。KMeans はクラスタ数が決まっている場合に最も扱いやすく、分類結果を簡潔に得たいときに向いている。Dendrogram はデータのまとまり方や階層構造を理解したいときに有効であり、クラスタ数を検討する段階でも役立つ。DBSCAN は密度構造や外れ値の存在を重視したい場合に適している。したがって、単純に 3 群へ分けたいなら KMeans、構造を観察したいなら Dendrogram、外れ値や不規則な分布を考えたいなら DBSCAN を使うのがよいと考えられる。
